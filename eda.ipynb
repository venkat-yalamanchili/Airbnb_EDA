```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('datasets.csv',encoding_errors='ignore')
```

```python
df.dropna(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# To see the duplicated records
df[df.duplicated()]
```

```python
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
df.dtypes
```

```python
# You can change the data types of columns
df['id'] = df['id'].astype(object) #Now id column is changed from float too object
df['host_id'] = df['host_id'].astype(object) 
df.dtypes
```

Univariate Analysis

```python
# Price distribution
sns.histplot(data=df, x='price', kde=True)
```

By seeing this we can know that there are some outliers

```python
#Identifying outliers in price
sns.boxplot(data=df, x='price')
```

```python
df1 = df[df['price']<1500] # Here we filtered out the price column <1500 only 
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df1, x='price', kde=True,bins=100)
plt.title('Price distribution')
plt.ylabel("Frequency")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df1, x='availability_365', kde=True)
plt.title('Availability distribution')
plt.ylabel("Frequency")
```

```python
# Pandas Group by
df1.groupby(by='neighbourhood_group')['price'].mean()
#df1.groupby(by='neighbourhood_group')['price'].agg(['min','mean','max'])
```

```python
#feature engineering
# We want a new column "Price per bed"
df1['price_per_bed'] = df1['price']/df1['beds']
```

```python
df1.head()
```

```python
df1.groupby(by='neighbourhood_group')['price_per_bed'].mean()
```

```python
#Bivariate Analysis
sns.barplot(data=df1,x='neighbourhood_group',y='price')
```
